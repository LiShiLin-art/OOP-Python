В этом мире сосуществуют обычные люди, **колдуны** (потомственные маги с врождённым даром) и **дивы** — опасные существа из параллельного мира **Пустоши**, способные менять облик и обладающие огромной силой[reference:1]. **Граф Гермес Аркадьевич Аверин** — потомственный колдун высшей категории, частный детектив. Умен, образован, обладает дворянской честью и профессиональной дотошностью[reference:3]. Его врождённое оружие — **хлёсткая Плеть**, но он предпочитает независимость и не любит бюрократию[reference:4].

**Див Кузя** — фамильяр Аверина, способный принимать облик то серого кота с глазами разного цвета, то рыжеволосого юноши[reference:5]. Добродушный, немного наивный, он создаёт тёплую и комическую атмосферу. В динамике отношений Аверина и Кузи происходит важнейшая трансформация: граф изначально воспринимает дивов как опасный инструмент, но благодаря Кузе его взгляды меняются[reference:6].

### Завязка сюжета

В Петербурге **пропадает див по имени Владимир** — бывший императорский див, известный своей строгостью и правильностью[reference:7]. Аверин берётся за расследование. Ему предстоит:

- опросить свидетелей в магическом Петербурге;
- найти улики в местах, связанных с Пустошью;
- разгадать загадку исчезновения;
- принять моральный выбор: спасти дива или раскрыть правду, которая может разрушить репутацию Империи.

### Игровой цикл (текстовая версия)

Игрок управляет Авериным. На каждом шаге:
1. Игра показывает **локацию** (особняк, Академия, Пустошь).

3. Игра реагирует: выдаёт **улику**, **диалог** или **событие**.
4. Игрок продвигается по сюжету, пока не раскроет тайну.

# 1. UML-диаграмма классов

## Теория: зачем UML и как его читать

**UML** (Unified Modeling Language) — стандартный язык визуального моделирования. Диаграмма классов показывает:
- **классы** (прямоугольники с тремя секциями);
- **атрибуты** (данные класса);
- **методы** (поведение класса);
- **связи** между классами.

**Модификаторы доступа:** `+` — публичный, `-` — приватный, `#` — защищённый.

**Основные связи:**

| Связь | Обозначение | Смысл | Пример |
|---|---|---|---|
| Наследование | полый треугольник | «является» | `Detective → Character` |
| Композиция | закрашенный ромб | «содержит (жизненный цикл)» | `Game ◆— Location` |
| Ассоциация | линия | «использует» | `Detective — Div` |

- Лутц М. «Изучаю Python», глава 26 (объектная модель).

# 2. Структура репозитория

## Теория: зачем нужна структура проекта

**Структура проекта** — это не формальность. Она:
- помогает быстро находить нужные файлы;
- разделяет ответственности (код, тесты, документация);
- облегчает командную работу;
- позволяет инструментам (pytest, mypy, flake8) работать автоматически.

**Стандартная структура для игры:**

```
mystic_detective/
├── core/           # базовые классы, интерфейсы, утилиты
│   ├── __init__.py
│   └── character.py   # Character, ABC
├── game/           # игровой цикл, управление
│   ├── __init__.py
│   └── game.py        # Game
├── entities/       # игровые сущности
│   ├── __init__.py
│   ├── detective.py   # Detective
│   ├── div.py         # Div
│   ├── location.py    # Location
│   └── clue.py        # Clue
├── ui/             # ввод/вывод
│   ├── __init__.py
│   └── ui.py          # UI, FakeUI
├── tests/          # тесты
│   ├── __init__.py
│   ├── test_detective.py
│   ├── test_div.py
│   └── test_game.py
├── docs/           # документация
│   ├── analysis.md    # анализ предметной области

├── main.py         # точка входа
├── requirements.txt
├── .gitignore
└── README.md
```

**Почему именно так:**
- `core/` — то, что не зависит от конкретной игры (базовые интерфейсы).
- `entities/` — игровые объекты (детектив, див, локация).
- `game/` — оркестрация (игровой цикл).
- `ui/` — изоляция ввода-вывода (можно заменить на GUI).
- `tests/` — автоматические тесты.
- `docs/` — документация (ADR, UML, анализ).

**Для углублённого изучения:**
- [Python Application Layouts](https://realpython.com/python-application-layouts/)
- [The Hitchhiker's Guide to Python: Structuring Your Project](https://docs.python-guide.org/writing/structure/)

````python
import os

PROJECT = 'mystic_detective'
for folder in ['core', 'game', 'entities', 'ui', 'tests', 'docs']:
    os.makedirs(os.path.join(PROJECT, folder), exist_ok=True)
    init_path = os.path.join(PROJECT, folder, '__init__.py')
    if not os.path.exists(init_path):
        with open(init_path, 'w', encoding='utf-8') as f:
            f.write(f'"""Пакет {folder}."""\n')

print("✅ Структура создана:")
for root, dirs, files in os.walk(PROJECT):
    level = root.replace(PROJECT, '').count(os.sep)
    indent = ' ' * 2 * level
    print(f"{indent}{os.path.basename(root)}/")
    subindent = ' ' * 2 * (level + 1)
    for file in files:
        print(f"{subindent}{file}")
````

# 3. Процедурный прототип (до рефакторинга)

## Теория: что такое процедурный код и почему его нужно рефакторить

**Процедурный код** организован вокруг **действий**: есть функции, которые работают с глобальными данными или передают данные через аргументы. Это работает для маленьких программ, но плохо масштабируется:
- данные разбросаны по переменным;
- функции знают слишком много о структуре данных;
- сложно тестировать отдельные части;
- изменение одной части ломает другие.

**Рефакторинг в ООП** превращает «список функций» в «набор взаимодействующих объектов». Каждый объект отвечает за своё состояние и поведение.

Ниже — процедурный прототип игры, который мы будем рефакторить. **Не запускайте его** — `input()` ждёт ввода. Просто прочитайте и проанализируйте.

````python
%%writefile mystic_detective/prototype_procedural.py
"""Процедурный прототип игры «Тайна исчезнувшего дива».

Только для анализа — не запускать (использует input).
"""

# Глобальные данные
locations = {
    'особняк': {
        'description': 'Старый особняк Авериных на Петроградке.',
        'clues': ['Странный запах гари'],
        'characters': ['Дворецкий'],
    },
    'академия': {
        'description': 'Академия колдовских наук.',
        'clues': ['Записка с планом Пустоши'],
        'characters': ['Ректор', 'Студент'],
    },
    'пустошь': {
        'description': 'Ледяная пустыня, мир дивов.',
        'clues': ['Следы Владимира'],
        'characters': ['Древний див'],
    },
}

player_energy = 100
found_clues = []
current_location = 'особняк'
kuzia_loyalty = 50


def show_location():
    loc = locations[current_location]
    print(f"\n=== {current_location.upper()} ===")
    print(loc['description'])


def search_location():
    loc = locations[current_location]
    if loc['clues']:
        clue = loc['clues'].pop(0)
        found_clues.append(clue)
        print(f"Найдена улика: {clue}")
    else:
        print("Здесь больше ничего нет.")


def talk_to_character():
    loc = locations[current_location]
    if loc['characters']:
        char = loc['characters'][0]
        print(f"{char} рассказывает что-то важное...")
        global kuzia_loyalty
        kuzia_loyalty += 5
    else:
        print("Здесь не с кем говорить.")


def move_to(location_name):
    global current_location, player_energy
    if location_name in locations:
        current_location = location_name
        player_energy -= 10
        print(f"Вы переместились в {location_name}. Энергия: {player_energy}")
    else:
        print("Такой локации нет.")


def check_victory():
    return len(found_clues) >= 3


def game_loop():
    print("=== ТАЙНА ИСЧЕЗНУВШЕГО ДИВА ===")
    while player_energy > 0:
        show_location()
        print("1. Осмотреть место")
        print("2. Поговорить")
        print("3. Переместиться")
        print("4. Выйти")
        choice = input("Выбор: ")
        if choice == '1':
            search_location()
        elif choice == '2':
            talk_to_character()
        elif choice == '3':
            loc_name = input("Куда? (особняк/академия/пустошь): ")
            move_to(loc_name)
        elif choice == '4':
            break
        if check_victory():
            print("Вы раскрыли тайну!")
            break


if __name__ == '__main__':
    game_loop()
````

### Анализ процедурного кода

| Что | Где | Куда перенести |
|---|---|---|
| `locations` | глобальный словарь | класс `Location` |
| `player_energy` | глобальная переменная | атрибут `Detective._energy` |
| `found_clues` | глобальный список | атрибут `Detective._clues` |
| `current_location` | глобальная переменная | атрибут `Game._current_location` |
| `kuzia_loyalty` | глобальная переменная | атрибут `Div._loyalty` |
| `show_location()`, `search_location()` | функции | методы `Location` и `Game` |
| `talk_to_character()` | функция | метод `Location` |
| `move_to()` | функция | метод `Game` |
| `check_victory()` | функция | метод `Game` |
| `input()`, `print()` | прямое использование | класс `UI` |

**Ключевые проблемы процедурного кода:**
1. **Глобальное состояние** — трудно тестировать, легко сломать.
2. **Смешение ответственностей** — `game_loop` знает и о локациях, и о вводе, и о победе.
3. **Нет инкапсуляции** — любая функция может изменить любую переменную.
4. **Сложно расширять** — добавление новой локации требует правок в нескольких местах.

# 4. ООП-версия: реализация классов

## Теория: ключевые принципы рефакторинга

1. **Одна ответственность на класс.** `Game` управляет циклом, `Location` хранит локацию, `UI` отвечает за ввод/вывод.
2. **Инкапсуляция.** Атрибуты приватные (`_name`), доступ через свойства (`@property`).
3. **Инъекция зависимостей.** `Game` и `Detective` получают `UI` через конструктор — это делает их тестируемыми.
4. **Композиция.** `Game` содержит `Detective` и `Location`; `Location` содержит `Clue`.

**Для углублённого изучения:**
- Martin Fowler. «Refactoring: Improving the Design of Existing Code».
- [Real Python: Object-Oriented Programming in Python](https://realpython.com/python3-object-oriented-programming/)
- Лутц М. «Изучаем Python», главы 25–27.

````python
%%writefile mystic_detective/core/character.py
"""Базовый класс для всех персонажей."""


class Character:
    """Базовый персонаж игры."""

    def __init__(self, name: str, description: str = ""):
        self._name = name
        self._description = description

    @property
    def name(self) -> str:
        return self._name

    @property
    def description(self) -> str:
        return self._description

    def introduce(self) -> str:
        """Краткое представление персонажа."""
        return f"{self._name}: {self._description}"

    def __str__(self) -> str:
        return self._name

    def __repr__(self) -> str:
        return f"{self.__class__.__name__}({self._name!r})"
````

````python
%%writefile mystic_detective/entities/clue.py
"""Улика — единица информации, найденная детективом."""


class Clue:
    """Улика: название, описание, важность."""

    def __init__(self, name: str, description: str = "", is_key: bool = False):
        self._name = name
        self._description = description
        self._is_key = is_key

    @property
    def name(self) -> str:
        return self._name

    @property
    def description(self) -> str:
        return self._description

    @property
    def is_key(self) -> bool:
        return self._is_key

    def __str__(self) -> str:
        marker = "[КЛЮЧЕВАЯ] " if self._is_key else ""
        return f"{marker}{self._name}"

    def __repr__(self) -> str:
        return f"Clue({self._name!r}, key={self._is_key})"
````

````python
%%writefile mystic_detective/entities/div.py
"""Див — магическое существо из Пустоши."""

from core.character import Character


class Div(Character):
    """Див: напарник детектива, способен менять облик."""

    FORMS = ["кот", "юноша", "ворон"]

    def __init__(self, name: str, form: str = "кот", loyalty: int = 50):
        super().__init__(name, "Див из Пустоши")
        if form not in self.FORMS:
            raise ValueError(f"Недопустимая форма: {form}. Допустимые: {self.FORMS}")
        self._form = form
        self._loyalty = max(0, min(100, loyalty))

    @property
    def form(self) -> str:
        return self._form

    @property
    def loyalty(self) -> int:
        return self._loyalty

    def transform(self, form: str) -> None:
        """Сменить облик."""
        if form not in self.FORMS:
            raise ValueError(f"Недопустимая форма: {form}")
        self._form = form

    def scout(self, location_name: str) -> str:
        """Разведка локации. Возвращает описание увиденного."""
        self._loyalty = min(100, self._loyalty + 5)
        return f"{self._name} в форме {self._form} разведал {location_name}"

    def speak(self) -> str:
        """Реплика дива."""
        return f"— {self._name}: Что-то здесь нечисто, Гермес Аркадьевич..."
````

````python
%%writefile mystic_detective/entities/detective.py
"""Детектив — главный герой."""

from core.character import Character
from entities.clue import Clue
from entities.div import Div


class Detective(Character):
    """Граф Аверин: колдун, детектив, владелец Плети."""

    def __init__(self, name: str, energy: int = 100, div: Div | None = None):
        super().__init__(name, "Частный детектив, колдун высшей категории")
        self._energy = energy
        self._max_energy = energy
        self._clues: list[Clue] = []
        self._div = div

    @property
    def energy(self) -> int:
        return self._energy

    @property
    def max_energy(self) -> int:
        return self._max_energy

    @property
    def clues(self) -> list[Clue]:
        return list(self._clues)

    @property
    def div(self) -> Div | None:
        return self._div

    def add_clue(self, clue: Clue) -> None:
        """Добавить улику в список найденных."""
        if clue not in self._clues:
            self._clues.append(clue)

    def investigate(self, location_name: str) -> str:
        """Осмотреть локацию. Тратит энергию."""
        if self._energy < 10:
            return "Не хватает энергии для осмотра."
        self._energy -= 10
        return f"{self._name} осматривает {location_name}"

    def use_whip(self) -> str:
        """Применить Плеть (врождённое оружие)."""
        if self._energy < 20:
            return "Не хватает энергии для Плети."
        self._energy -= 20
        return "Плеть рассекает воздух!"

    def rest(self, amount: int = 20) -> None:
        """Отдохнуть и восстановить энергию."""
        self._energy = min(self._max_energy, self._energy + amount)

    def __str__(self) -> str:
        return f"{self._name} (энергия: {self._energy}/{self._max_energy})"
````

````python
%%writefile mystic_detective/entities/location.py
"""Локация — место, где происходят события."""

from entities.clue import Clue


class Location:
    """Локация: описание, улики, персонажи."""

    def __init__(self, name: str, description: str = ""):
        self._name = name
        self._description = description
        self._clues: list[Clue] = []
        self._characters: list[str] = []

    @property
    def name(self) -> str:
        return self._name

    @property
    def description(self) -> str:
        return self._description

    @property
    def clues(self) -> list[Clue]:
        return list(self._clues)

    @property
    def characters(self) -> list[str]:
        return list(self._characters)

    def add_clue(self, clue: Clue) -> None:
        self._clues.append(clue)

    def add_character(self, character_name: str) -> None:
        self._characters.append(character_name)

    def search(self) -> Clue | None:
        """Найти улику. Возвращает Clue или None."""
        if self._clues:
            return self._clues.pop(0)
        return None

    def get_character(self) -> str | None:
        """Получить персонажа для диалога."""
        if self._characters:
            return self._characters[0]
        return None

    def __str__(self) -> str:
        return self._name
````

````python
%%writefile mystic_detective/ui/ui.py
"""Интерфейс пользователя."""


class UI:
    """Абстракция ввода/вывода."""

    def show_message(self, message: str) -> None:
        print(message)

    def get_input(self, prompt: str) -> str:
        return input(prompt)


class FakeUI(UI):
    """UI для тестов и демонстраций: заранее заданные ответы."""

    def __init__(self, answers: list[str]):
        self._answers = iter(answers)
        self.messages: list[str] = []

    def get_input(self, prompt: str) -> str:
        return next(self._answers)

    def show_message(self, message: str) -> None:
        self.messages.append(message)
        print(message)
````

````python
%%writefile mystic_detective/game/game.py
"""Игровой цикл."""

from entities.detective import Detective
from entities.div import Div
from entities.location import Location
from entities.clue import Clue
from ui.ui import UI


class Game:
    """Управляет игровым циклом «Тайна исчезнувшего дива»."""

    def __init__(self, ui: UI | None = None):
        self._ui = ui if ui is not None else UI()
        self._div = Div("Кузя", form="кот", loyalty=50)
        self._detective = Detective("Граф Аверин", energy=100, div=self._div)
        self._locations: dict[str, Location] = self._build_locations()
        self._current_location: Location = self._locations['особняк']

    def _build_locations(self) -> dict[str, Location]:
        """Создать все локации игры."""
        loc1 = Location("особняк", "Старый особняк Авериных на Петроградке.")
        loc1.add_clue(Clue("Странный запах гари", "Пахнет серой и гарью", is_key=False))
        loc1.add_character("Дворецкий")

        loc2 = Location("академия", "Академия колдовских наук.")
        loc2.add_clue(Clue("Записка с планом Пустоши", "Кто-то рисовал карту", is_key=True))
        loc2.add_character("Ректор")
        loc2.add_character("Студент")

        loc3 = Location("пустошь", "Ледяная пустыня, мир дивов.")
        loc3.add_clue(Clue("Следы Владимира", "Отпечатки лап на льду", is_key=True))
        loc3.add_character("Древний див")

        return {"особняк": loc1, "академия": loc2, "пустошь": loc3}

    @property
    def detective(self) -> Detective:
        return self._detective

    @property
    def current_location(self) -> Location:
        return self._current_location

    def run(self) -> None:
        """Запустить игровой цикл."""
        self._ui.show_message("=== ТАЙНА ИСЧЕЗНУВШЕГО ДИВА ===")
        self._ui.show_message("Вы — граф Гермес Аверин. Див Владимир пропал.")

        while self._detective.energy > 0:
            self._show_status()
            choice = self._ui.get_input(
                "\n1. Осмотреть место\n2. Поговорить\n3. Переместиться\n4. Выйти\nВыбор: "
            )

            if choice == '1':
                self._search()
            elif choice == '2':
                self._talk()
            elif choice == '3':
                self._move()
            elif choice == '4':
                break

            if self.check_victory():
                self._ui.show_message("\nВы раскрыли тайну исчезновения Владимира!")
                return

        self._ui.show_message("\nЭнергия закончилась. Расследование провалено.")

    def _show_status(self) -> None:
        self._ui.show_message(f"\n=== {self._current_location.name.upper()} ===")
        self._ui.show_message(self._current_location.description)
        self._ui.show_message(f"{self._detective}")
        if self._div:
            self._ui.show_message(f"Кузя (форма: {self._div.form}, лояльность: {self._div.loyalty})")

    def _search(self) -> None:
        self._ui.show_message(self._detective.investigate(self._current_location.name))
        clue = self._current_location.search()
        if clue:
            self._detective.add_clue(clue)
            self._ui.show_message(f"Найдена улика: {clue}")
        else:
            self._ui.show_message("Здесь больше ничего нет.")

    def _talk(self) -> None:
        character = self._current_location.get_character()
        if character:
            self._ui.show_message(f"{character} рассказывает что-то важное...")
            if self._div:
                self._ui.show_message(self._div.speak())
        else:
            self._ui.show_message("Здесь не с кем говорить.")

    def _move(self) -> None:
        names = "/".join(self._locations.keys())
        loc_name = self._ui.get_input(f"Куда? ({names}): ")
        if loc_name in self._locations:
            self._current_location = self._locations[loc_name]
            self._detective.rest(5)
            self._ui.show_message(f"Вы переместились в {loc_name}.")
        else:
            self._ui.show_message("Такой локации нет.")

    def check_victory(self) -> bool:
        """Победа: найдены обе ключевые улики."""
        key_clues = [c for c in self._detective.clues if c.is_key]
        return len(key_clues) >= 2
````

````python
%%writefile mystic_detective/main.py
"""Точка входа в игру."""

import sys
import os
sys.path.insert(0, os.path.dirname(os.path.abspath(__file__)))

from game.game import Game


def main() -> None:
    Game().run()


if __name__ == "__main__":
    main()
````

# 5. Юнит-тесты

## Теория: зачем тестировать игровую логику

Юнит-тесты проверяют **отдельные классы** в изоляции. В игре особенно важно тестировать:
- **состояние** (энергия, лояльность, список улик);
- **переходы состояний** (смена локации, победа);
- **граничные случаи** (энергия = 0, пустой список улик).

**Mock** позволяет изолировать UI: в тестах мы не хотим ждать ввода, поэтому подставляем `Mock` вместо реального `UI`.

**Для углублённого изучения:**
- [pytest documentation](https://docs.pytest.org/)
- [Real Python: Testing Your Code with pytest](https://realpython.com/pytest-python-testing/)

````python
%%writefile mystic_detective/tests/test_detective.py
"""Тесты класса Detective."""

from entities.detective import Detective
from entities.clue import Clue
from entities.div import Div


class TestDetective:
    def test_initial_state(self):
        d = Detective("Аверин")
        assert d.energy == 100
        assert d.clues == []

    def test_investigate_reduces_energy(self):
        d = Detective("Аверин", energy=100)
        d.investigate("особняк")
        assert d.energy == 90

    def test_use_whip_reduces_energy(self):
        d = Detective("Аверин", energy=100)
        d.use_whip()
        assert d.energy == 80

    def test_add_clue(self):
        d = Detective("Аверин")
        clue = Clue("Записка", is_key=True)
        d.add_clue(clue)
        assert len(d.clues) == 1
        assert d.clues[0].is_key

    def test_add_duplicate_clue(self):
        d = Detective("Аверин")
        clue = Clue("Записка")
        d.add_clue(clue)
        d.add_clue(clue)
        assert len(d.clues) == 1

    def test_rest_restores_energy(self):
        d = Detective("Аверин", energy=50)
        d.rest(30)
        assert d.energy == 80

    def test_rest_caps_at_max(self):
        d = Detective("Аверин", energy=50)
        d.rest(999)
        assert d.energy == 100
````

````python
%%writefile mystic_detective/tests/test_div.py
"""Тесты класса Div."""

from entities.div import Div
import pytest


class TestDiv:
    def test_initial_form(self):
        k = Div("Кузя", form="кот")
        assert k.form == "кот"

    def test_invalid_form(self):
        with pytest.raises(ValueError):
            Div("Кузя", form="дракон")

    def test_transform(self):
        k = Div("Кузя")
        k.transform("юноша")
        assert k.form == "юноша"

    def test_scout_increases_loyalty(self):
        k = Div("Кузя", loyalty=50)
        k.scout("академия")
        assert k.loyalty == 55

    def test_loyalty_capped_at_100(self):
        k = Div("Кузя", loyalty=98)
        k.scout("академия")
        assert k.loyalty == 100
````

````python
%%writefile mystic_detective/docs/adr_001.md
# ADR-001: Разделение Game и UI

## Контекст
Процедурный прототип использовал input() и print() напрямую, что делало невозможным тестирование
и замену консольного интерфейса на GUI.

## Решение
Выделен класс UI, инкапсулирующий ввод/вывод. Game и Detective получают UI через конструктор
(инъекция зависимостей).

## Рассмотренные альтернативы
1. Глобальный UI-модуль — проще, но сложнее тестировать.
2. Наследование — приводит к жёсткой связи.
3. Инъекция через конструктор (выбрано) — гибко, тестируемо.

## Последствия
- ✅ Логика тестируется через Mock/FakeUI.
- ✅ Легко заменить консоль на GUI.
- ⚠️ +1 уровень косвенности.
````
