# 1. Зачем ООП в разработке игр

## 1.1. Специфика игровой разработки

Игра — это, пожалуй, самый сложный тип программного обеспечения после операционных систем. В одном кадре (обычно 16 миллисекунд при 60 FPS) должны уложиться:

- обработка пользовательского ввода (клавиатура, мышь, геймпад);
- обновление физики (коллизии, гравитация, перемещения);
- работа искусственного интеллекта (враги, NPC, поведение);
- проигрывание звука и музыки;
- отрисовка графики;
- сетевая синхронизация (в мультиплеере);
- игровая логика (правила, состояния, квесты).

Все эти подсистемы должны работать **синхронно**, **производительно** и — главное — **расширяемо**. Добавление нового типа врага, оружия или уровня не должно требовать переписывания половины кода.

## 1.2. Почему процедурный подход плохо масштабируется

Процедурный код для игры выглядит естественно на старте: есть переменные `player_x`, `player_hp`, есть функции `move_player()`, `damage_player()`. Но представьте, что таких сущностей десятки, у каждой — своё состояние и свои правила. Появляются глобальные переменные, гигантские функции `update_everything()`, а изменение одной системы ломает три других.

**Пример деградации процедурного кода:**

```python
# Начало проекта — всё просто
player_x = 0
player_hp = 100

def move_player(dx):
    global player_x
    player_x += dx

# Через месяц — здесь 50 глобальных переменных
# и функции, которые зависят от всех сразу
```

С ростом проекта это превращается в **спагетти-код**: сложно понять, сложно тестировать, невозможно переиспользовать.

## 1.3. Как ООП решает эти проблемы

Игровой мир **естественным образом** состоит из объектов: персонажей, врагов, снарядов, препятствий, предметов, камер, таймеров. Каждый объект обладает:

- **состоянием** (координаты, здоровье, скорость);
- **поведением** (двигаться, атаковать, получать урон, исчезать);
- **идентичностью** (это конкретный персонаж, а не абстрактный «набор полей»).

ООП позволяет описать эти сущности классами и организовать их взаимодействие. Вместо «глобальной кучи переменных» появляются автономные объекты, каждый из которых сам отвечает за своё состояние.

## 1.4. ООП как способ мышления

**Важно понять:** ООП — это не синтаксис, а **способ мышления**. Вместо вопроса «какие функции мне нужно написать» вы задаёте вопросы:

- Какие сущности существуют в моей игре?
- Что каждая из них знает (состояние)?
- Что каждая из них умеет (поведение)?
- Как они взаимодействуют?

Этот переход от «функционального» к «объектному» мышлению — ключевая задача данного модуля. Студенты, которые механически добавляют `class` вокруг старых функций, не получают преимуществ ООП. Настоящий ООП начинается с **декомпозиции предметной области**.

## 1.5. ООП в Python: гибкость и компромиссы

Python реализует ООП особенно гибко. В отличие от Java или C++, где классы — практически единственный способ организации кода, Python позволяет смешивать процедурный, функциональный и объектный стили. Это означает, что вы можете **постепенно** переходить от процедурного кода к объектному — что особенно ценно при рефакторинге прототипов.

Однако гибкость имеет цену: Python не заставляет вас соблюдать дисциплину. Инкапсуляция — «джентльменское соглашение», множественное наследование может запутать, а динамическая типизация — скрыть ошибки. Именно поэтому курс делает акцент не только на синтаксисе, но и на **практиках проектирования**.

# 2. Классы и объекты

## 2.1. Что такое класс

**Класс** — это шаблон (чертёж, blueprint), описывающий структуру и поведение будущих объектов. В Python класс создаётся ключевым словом `class`:

```python
class Player:
    """Класс, описывающий игрока."""
    pass
```

Даже такой минимальный класс — уже полноценный **тип**. Создание нового класса создаёт новый тип объекта, позволяя создавать новые экземпляры этого типа.

**Важнейшее различие:** `Player` — это не игрок. Это описание того, каким может быть игрок. Класс — категория, объект — конкретный представитель категории.

**Аналогия из жизни:**  
Класс `Player` — это как чертёж автомобиля на заводе. Сам чертёж — не автомобиль, но по нему можно собрать много одинаковых машин. Каждая машина будет иметь свои номера, свой пробег, свой цвет — но все они построены по одному чертежу.

## 2.2. Что такое объект

**Объект** (экземпляр, instance) — это конкретная сущность, созданная по шаблону класса:

```python
player1 = Player()
player2 = Player()
```

`player1` и `player2` — два разных объекта одного класса. Изменение состояния одного не влияет на другой.

**Практический пример:**  
В игре «Космический шутер» класс `Player` описывает корабль игрока, `Enemy` — вражеский корабль. При запуске игры создаётся один объект `Player` и множество `Enemy`. У каждого — свои координаты, здоровье, скорость.

## 2.3. «Всё есть объект»

В Python **всё является объектом**: числа, строки, списки, функции, модули и даже сами классы. Класс `Player` — тоже объект (экземпляр метакласса `type`).

Это означает, что с классами можно работать как с любыми другими объектами: передавать в функции, хранить в переменных, создавать динамически:

```python
enemy_class = Enemy
enemy1 = enemy_class()  # создать объект через переменную
```

Этот приём используется в паттернах проектирования (Factory, Registry), которые будут изучены в модуле 6.

## 2.4. Проверка типов

Python предоставляет инструменты для работы с типами:

```python
type(player1)         # <class 'Player'>
isinstance(player1, Player)  # True
```

**Важно:** в Python принято использовать `isinstance()` вместо `type() is Class`. Первый вариант учитывает наследование (объект подкласса тоже является объектом родителя), второй — нет.

## 2.5. Класс как фабрика объектов

Когда вы пишете `Player()`, Python выполняет два шага:
1. Создаёт новый экземпляр (`__new__`).
2. Инициализирует его (`__init__`).

Обычно нас интересует второй шаг — именно там задаётся начальное состояние объекта.

**Для углублённого изучения:**
- [Princeton CS: Objects and Classes](https://www.cs.princeton.edu/courses/archive/fall20/cos126/)
- [CMU 15-112: OOP Notes](https://www.cs.cmu.edu/~112/)

````python
class Player:
    """Класс, описывающий игрока."""
    pass

# Создание двух независимых объектов
player1 = Player()
player2 = Player()

print(type(player1))
print(player1 is player2)
print(isinstance(player1, Player))
````

**Отметка для особого внимания:**  
В Python «всё есть объект»: числа, строки, функции, модули и сами классы. Проверим:

````python
print(type(42))
print(type(str))
print(type(Player))  # метакласс type
````

# 3. Атрибуты экземпляра и класса

## 3.1. Атрибуты экземпляра

**Атрибуты экземпляра** — данные, принадлежащие конкретному объекту. Они создаются в методах экземпляра, чаще всего в `__init__`, с использованием `self`:

```python
self.name = name
self.health = health
```

Каждый объект имеет **собственную копию** таких атрибутов. Изменение `player1.health` не затрагивает `player2.health`.

**Когда использовать:** для состояния, которое уникально для каждого объекта — координаты, текущее здоровье, имя, инвентарь.

## 3.2. Атрибуты класса

**Атрибуты класса** — данные, принадлежащие самому классу и разделяемые всеми экземплярами. Определяются в теле класса, вне методов:

```python
class Enemy:
    max_health = 100      # атрибут класса
    enemy_count = 0       # атрибут класса (счётчик)
```

**Когда использовать:** для констант, общих для всех экземпляров (лимиты, гравитация, цвет по умолчанию), а также для счётчиков созданных объектов.

## 3.3. Разница и типичная ловушка

**Ключевое отличие:** атрибуты экземпляра ищутся в `obj.__dict__`, атрибуты класса — в `Class.__dict__`. При обращении `obj.attr` Python сначала ищет в экземпляре, потом в классе, потом в родителях.

**Ловушка:** если вы присвоите значение через экземпляр (`enemy1.max_health = 200`), Python создаст **атрибут экземпляра** с тем же именем, который «затенит» атрибут класса. Сам атрибут класса не изменится. Это одна из самых частых ошибок у начинающих.

**Правило:** если нужно изменить атрибут класса — используйте `ClassName.attr = value`, а не `obj.attr = value`.

## 3.4. Проверка через `__dict__`

Атрибуты экземпляра хранятся в `obj.__dict__` — это словарь. Атрибуты класса — в `Class.__dict__`. Это позволяет явно проверить, где находится атрибут.

**Для углублённого изучения:**
- [Real Python: Class and Instance Attributes](https://realpython.com/python-class-attributes/)
- [Python mailing list: class vs instance attributes](https://mail.python.org/pipermail/tutor/)

````python
class Enemy:
    max_health = 100          # атрибут класса
    enemy_count = 0           # атрибут класса (счётчик)

    def __init__(self, name, health=None):
        self.name = name                              # атрибут экземпляра
        self.health = health or Enemy.max_health      # атрибут экземпляра
        Enemy.enemy_count += 1

e1 = Enemy("Goblin")
e2 = Enemy("Orc", 150)

print(f"Всего врагов: {Enemy.enemy_count}")
print(f"{e1.name}: {e1.health}, {e2.name}: {e2.health}")
print(f"max_health у класса: {Enemy.max_health}")
print(f"__dict__ e1: {e1.__dict__}")
print(f"__dict__ Enemy: {[k for k in Enemy.__dict__ if not k.startswith('__')]}")
````

**⚠️ Ловушка затенения:** присвоение через экземпляр создаёт новый атрибут экземпляра, а не изменяет атрибут класса:

````python
e1.max_health = 200
print(f"e1.max_health: {e1.max_health} (атрибут экземпляра)")
print(f"e2.max_health: {e2.max_health} (атрибут класса)")
print(f"Enemy.max_health: {Enemy.max_health} (не изменился)")
print(f"__dict__ e1 теперь: {e1.__dict__}")
````

# 4. Методы, `self` и передача сообщений

## 4.1. Что такое метод

**Метод** — это функция, определённая внутри класса. Методы описывают **поведение** объектов.

В Python есть три типа методов:
- **Методы экземпляра** — принимают `self`, работают с конкретным объектом (90% случаев).
- **Методы класса** (`@classmethod`) — принимают `cls`, работают с классом в целом.
- **Статические методы** (`@staticmethod`) — не принимают ни `self`, ни `cls`, логически связаны с классом, но не зависят от его состояния.

В этом модуле мы фокусируемся на методах экземпляра.

## 4.2. Что такое `self`

**`self`** — это ссылка на текущий экземпляр, для которого вызывается метод. Когда вы пишете `player1.take_damage(10)`, Python автоматически передаёт `player1` в качестве первого аргумента `self`.

**Важно понимать:** `self` — это **не ключевое слово**, а соглашение. Технически можно назвать первый аргумент как угодно (`this`, `obj`, `me`), но `self` — общепринятый стандарт, которого придерживается всё сообщество Python. Отклонение от него считается плохим стилем.

**Практическое объяснение:**  
Представьте, что класс — это чертёж дома, а `self` — это конкретный дом, к которому вы подошли. Метод `paint_wall(self, color)` означает «покрасить стену **этого** дома в цвет `color`». Без `self` метод не знал бы, какую стену красить.

## 4.3. Передача сообщений

В ООП взаимодействие объектов часто описывается как **передача сообщений**. Когда объект A вызывает метод объекта B, говорят, что A отправляет сообщение B.

**Принцип проектирования:** объекты должны взаимодействовать через **чётко определённые интерфейсы** (публичные методы), а не через прямое манипулирование внутренними данными. Это основа инкапсуляции.

**Пример из игры:**  
Оружие (`Weapon`) не должно напрямую уменьшать `target.health`. Оно должно вызвать `target.take_damage(amount)`. Тогда цель сама решает, как обработать урон (учесть броню, вызвать смерть, воспроизвести анимацию).

## 4.4. Почему `self` явный

В отличие от Java или C++, где `this` — неявная ссылка, в Python `self` всегда явный. Это сделано намеренно:
- улучшает читаемость — всегда видно, что атрибут принадлежит объекту;
- упрощает понимание области видимости;
- позволяет использовать свободные функции и методы единообразно.

**Для углублённого изучения:**
- [Python mailing list: Understanding self](https://mail.python.org/pipermail/tutor/)
- [Real Python: Python's self](https://realpython.com/python-self/)

````python
class Weapon:
    def __init__(self, damage):
        self.damage = damage

    def attack(self, target):
        """Отправить 'сообщение' цели."""
        print(f"Атакуем {target.name} на {self.damage} урона")
        target.take_damage(self.damage)


class Enemy:
    def __init__(self, name, health):
        self.name = name
        self.health = health

    def take_damage(self, amount):
        self.health -= amount
        print(f"{self.name} получил {amount}, осталось {self.health}")


sword = Weapon(15)
goblin = Enemy("Goblin", 50)
sword.attack(goblin)
````

# 5. Конструктор `__init__`

## 5.1. Роль `__init__`

**`__init__`** — специальный метод, автоматически вызываемый при создании объекта. Его задача — **инициализировать** объект: установить начальные значения атрибутов, выполнить настройку.

**Техническая тонкость:** `__init__` — это **не конструктор** в строгом смысле. Настоящий конструктор — метод `__new__`, который создаёт объект. `__init__` лишь инициализирует уже созданный объект. Для практических целей это различие редко важно, но полезно знать, что `__init__` возвращает `None` и не может заменить объект.

## 5.2. Аргументы по умолчанию

`__init__` может принимать аргументы с значениями по умолчанию:

```python
def __init__(self, name, health=100, speed=1.0):
    ...
```

Это удобно для создания объектов с разумными настройками «по умолчанию», но требует осторожности.

## 5.3. Опасность изменяемых значений по умолчанию

**Классическая ошибка Python:** использование изменяемых объектов (списков, словарей, множеств) как значений по умолчанию. Они создаются **один раз** при определении функции и **разделяются** между всеми вызовами.

```python
# ПЛОХО
def __init__(self, items=[]):
    self.items = items
```

Если создать два объекта, а затем добавить элемент в инвентарь первого, он появится и у второго — потому что оба ссылаются на один и тот же список.

**Правильное решение:** использовать `None` как значение по умолчанию и создавать новый объект внутри:

```python
def __init__(self, items=None):
    self.items = items if items is not None else []
```

Это правило особенно критично в игровых проектах, где инвентари, списки врагов и другие коллекции создаются часто.

## 5.4. Валидация входных данных

`__init__` — хорошее место для проверки корректности аргументов. Если здоровье не может быть отрицательным, проверьте это сразу:

```python
if health < 0:
    raise ValueError("Health cannot be negative")
```

Это принцип **fail fast**: ошибка обнаруживается в момент создания объекта, а не через час после начала игры.

- [Real Python: Python __init__](https://realpython.com/python-class-constructor/)
- [Python docs: __init__](https://docs.python.org/3/reference/datamodel.html#object.__init__)

````python
class Bullet:
    def __init__(self, x, y, speed=5.0, damage=10):
        self.x = x
        self.y = y
        self.speed = speed
        self.damage = damage
        self.active = True

    def update(self, dt):
        self.x += self.speed * dt


b = Bullet(0, 0)
b.update(1.0)
print(b.x)
````

**⚠️ Опасность изменяемых значений по умолчанию:**

````python
# ПЛОХО: изменяемый default
class BadInventory:
    def __init__(self, items=[]):
        self.items = items

inv1 = BadInventory()
inv2 = BadInventory()
inv1.items.append("sword")
print("Плохой вариант:", inv2.items)  # ['sword'] — неожиданно!

# ХОРОШО: None как default
class GoodInventory:
    def __init__(self, items=None):
        self.items = items if items is not None else []

inv3 = GoodInventory()
inv4 = GoodInventory()
inv3.items.append("sword")
print("Хороший вариант:", inv4.items)  # []
````

# 6. Инкапсуляция: `_` и `__`

## 6.1. Что такое инкапсуляция

**Инкапсуляция** — принцип, согласно которому внутреннее состояние объекта должно быть **скрыто** от внешнего мира, а доступ к нему — контролироваться через публичный интерфейс (методы). Это позволяет:
- защитить данные от некорректного изменения;
- скрыть детали реализации;
- изменить внутреннее устройство класса без нарушения внешнего кода.

## 6.2. Инкапсуляция в Python: не запрет, а соглашение

В Python **нет строгих модификаторов доступа** (как `private` в Java). Вместо этого используется соглашение об именовании:

| Префикс | Соглашение | Уровень |
|---|---|---|
| `name` | публичный | доступен всем |
| `_name` | защищённый (protected) | «не трогайте без необходимости» |
| `__name` | приватный (private) | name mangling |
| `__name__` | магический | внутренний для Python |

**Философия Python:** «Мы все взрослые люди». Программист должен сам понимать, что `_name` — внутренняя деталь, и не использовать её снаружи. Это **джентльменское соглашение**, а не запрет.

## 6.3. Соглашение `_` (одинарное подчёркивание)

Атрибуты и методы с одним подчёркиванием считаются **защищёнными**. Это сигнал: «не трогайте это снаружи класса». Python не запрещает доступ — просто сообщество считает это дурным тоном.

**Когда использовать:** для внутренних атрибутов, которые не должны изменяться напрямую, но нужны для работы методов.

## 6.4. Соглашение `__` (двойное подчёркивание) и name mangling

Атрибуты с двумя подчёркиваниями в начале (но не более двух в конце) подвергаются **name mangling** — Python автоматически переименовывает их, добавляя префикс `_ClassName`. То есть `__balance` в классе `BankAccount` превращается в `_BankAccount__balance`.

Это делает **случайный** доступ извне затруднительным, но не невозможным. Смысл — защита от конфликтов имён при наследовании и от случайных ошибок.

**Важно:** name mangling — не механизм безопасности. Это инструмент для избежания конфликтов имён. Не используйте `__` для «настоящей» защиты данных — для этого в Python просто нет механизма.

## 6.5. Практический пример: компонент здоровья

Рассмотрим класс `Health` — типичный компонент в игровом движке. Он инкапсулирует текущее и максимальное здоровье, предоставляя публичные методы `take_damage()` и `heal()`, а также свойства `current`, `max`, `is_alive`.

Пользователь класса не может напрямую присвоить `health.current = 999` — только через методы, которые проверяют границы.

**Для углублённого изучения:**
- [COMP70050: Encapsulation in Python](https://www.doc.ic.ac.uk/~svb/)
- [Real Python: Private Attributes in Python](https://realpython.com/python-private-attributes/)

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance    # приватный

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount

    def get_balance(self):
        return self.__balance


acc = BankAccount(1000)
acc.deposit(500)
print(acc.get_balance())

try:
    print(acc.__balance)
except AttributeError as e:
    print("Ошибка:", e)

# Обход (не рекомендуется):
print(acc._BankAccount__balance)
````

# 7. Свойства `@property`

## 7.1. Проблема: как контролировать доступ к атрибутам

В классическом ООП (Java, C++) для доступа к приватным полям используются геттеры и сеттеры:

```java
public int getHealth() { return health; }
public void setHealth(int h) { health = h; }
```

В Python такой подход считается **непитоничным** — он загромождает код. Вместо этого используется механизм **свойств** (`property`), который сохраняет синтаксис прямого доступа `obj.attr`, но позволяет выполнять произвольную логику при чтении и записи.

## 7.2. Декоратор `@property`

Свойство создаётся через декораторы:

```python
@property
def celsius(self):       # геттер
    return self._celsius

@celsius.setter
def celsius(self, value):  # сеттер
    if value < -273.15:
        raise ValueError(...)
    self._celsius = value
```

Теперь снаружи можно писать `temp.celsius = 30`, и Python автоматически вызовет сеттер.

## 7.3. Ключевые преимущества `@property`

1. **Синтаксис прямого доступа** — `temp.celsius` вместо `temp.get_celsius()`.
2. **Валидация при записи** — можно проверить значение перед сохранением.
3. **Вычисляемые свойства** — например, `fahrenheit` не хранится, а вычисляется на лету.
4. **Обратная совместимость** — если вы начали с публичного атрибута и позже решили добавить логику, вы можете превратить его в свойство без изменения внешнего кода.

## 7.4. Когда НЕ использовать `@property`

**Отметка для особого внимания:** не стоит превращать каждый атрибут в свойство «на всякий случай». Это замедляет код и загромождает класс. Используйте свойства только тогда, когда действительно нужна логика при доступе:
- валидация;
- вычисление производных значений;
- ленивая инициализация;
- обратная совместимость с API.

Если атрибут просто хранит значение — оставьте его публичным.

## 7.5. Практический пример: игровой персонаж с свойствами

Класс `Character` с приватными `_health`, `_mana` и публичными свойствами `health`, `mana`, `health_percent`. Сеттеры ограничивают значения в допустимых границах, а `health_percent` — вычисляемое свойство для UI.

**Для углублённого изучения:**
- [Real Python: Python's property()](https://realpython.com/python-property/)
- [Google Python Style Guide](https://google.github.io/styleguide/pyguide.html)

````python
class Health:
    def __init__(self, max_health=100):
        self._max = max_health
        self._current = max_health

    @property
    def current(self):
        return self._current

    @current.setter
    def current(self, value):
        self._current = max(0, min(value, self._max))

    @property
    def percentage(self):
        return self._current / self._max * 100

    @property
    def is_alive(self):
        return self._current > 0


hp = Health(120)
hp.current -= 30
print(hp.current, hp.percentage, hp.is_alive)
hp.current = 999
print(hp.current)  # 120 — ограничено сверху
````

# 8. Наследование и `super()`

## 8.1. Что такое наследование

**Наследование** — механизм, позволяющий создать новый класс на основе существующего. Подкласс наследует атрибуты и методы родителя и может добавлять свои или переопределять унаследованные.

**Отношение «является» (is-a):**  
`Player` **является** `GameObject`. `Enemy` **является** `GameObject`. Это означает, что везде, где ожидается `GameObject`, можно использовать `Player` или `Enemy` (принцип подстановки Барбары Лисков, который будет изучен в модуле 3).

## 8.2. Зачем нужно наследование

Наследование решает проблему **дублирования кода**. Вместо того чтобы писать `move()` и `draw()` в каждом классе, мы выносим их в общий базовый класс. Это соответствует принципу **DRY** (Don't Repeat Yourself).

## 8.3. Функция `super()`

`super()` возвращает временный объект-прокси, позволяющий вызывать методы родительского класса. Чаще всего используется для вызова `__init__` родителя:

```python
super().__init__(name, health)
```

**Почему `super()` лучше прямого вызова `BaseEnemy.__init__(self, ...)`?**  
При множественном наследовании `super()` следует порядку разрешения методов (MRO) и гарантирует, что каждый родительский класс будет вызван ровно один раз. Прямой вызов может привести к дублированию вызовов или пропуску классов.

## 8.4. Переопределение методов

Подкласс может **переопределить** метод родителя, предоставив свою реализацию. При этом часто вызывают родительскую версию через `super().method()`:

```python
class MagicWeapon(Weapon):
    def attack(self, target):
        print("Casting spell...")
        super().attack(target)
```

Это называется **расширением поведения**: мы не заменяем, а дополняем родительскую логику.

## 8.5. Множественное наследование

Python поддерживает **множественное наследование** — класс может наследоваться от нескольких родителей:

```python
class Character(Movable, Damageable):
    ...
```

**Внимание:** множественное наследование — мощный, но опасный инструмент. Оно может привести к сложным иерархиям и конфликтам методов (проблема «ромба»). В большинстве случаев предпочтительнее **композиция** (см. раздел 12). Множественное наследование оправдано для **миксинов** — небольших классов, добавляющих одну конкретную функциональность.

**Для углублённого изучения:**
- [Real Python: Object Inheritance](https://realpython.com/inheritance-composition-python/)
- [Python docs: super()](https://docs.python.org/3/library/functions.html#super)

````python
class GameObject:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def move(self, dx, dy):
        self.x += dx
        self.y += dy

    def draw(self):
        print(f"Object at ({self.x},{self.y})")


class Player(GameObject):
    def __init__(self, x, y, name):
        super().__init__(x, y)
        self.name = name
        self.health = 100

    def draw(self):
        print(f"Player {self.name} at ({self.x},{self.y})")


p = Player(10, 20, "Hero")
p.move(5, 5)
p.draw()
print(isinstance(p, GameObject))
````

**Множественное наследование:**

````python
class Movable:
    def move(self, dx, dy):
        self.x += dx
        self.y += dy

class Damageable:
    def take_damage(self, amount):
        self.health -= amount

class Character(Movable, Damageable):
    def __init__(self, x, y, health):
        self.x, self.y, self.health = x, y, health

c = Character(0, 0, 100)
c.move(3, 4)
c.take_damage(20)
print(c.x, c.y, c.health)
````

# 9. Полиморфизм

## 9.1. Что такое полиморфизм

**Полиморфизм** — способность объектов с разными типами реагировать на один и тот же вызов метода по-разному. В Python полиморфизм реализуется через **утиную типизацию** (duck typing): «Если это выглядит как утка, плавает как утка и крякает как утка, то это, вероятно, и есть утка».

Это означает, что Python **не проверяет тип** объекта при вызове метода. Если у объекта есть нужный метод — вызов сработает. Это делает код максимально гибким.

## 9.2. Полиморфизм через общий интерфейс

Классический пример: разные типы врагов, каждый со своим `attack()`. Цикл, обрабатывающий список врагов, не знает и **не должен знать** конкретные типы — он просто вызывает `attack()`.

```python
for enemy in enemies:
    enemy.attack(hero)
```

Это делает код **расширяемым**: добавление нового типа врага не требует изменения существующего кода.

## 9.3. Полиморфизм через наследование vs утиную типизацию

В Java/C++ полиморфизм требует **общего родителя** или **интерфейса**. В Python это не обязательно — достаточно, чтобы у объектов был одинаковый метод. Это называется **неявным полиморфизмом**.

Однако в больших проектах всё же рекомендуется определять общий интерфейс (через ABC или `Protocol`), чтобы IDE и линтеры могли проверять типы. Это будет изучено в модуле 2.

## 9.4. Практический пример: враги с разным поведением

Классы `MeleeEnemy`, `RangedEnemy`, `MagicEnemy` наследуются от `Enemy` и переопределяют `attack()`. При вызове `enemy.attack(hero)` каждый ведёт себя по-своему: рукопашный бьёт мечом, стрелок стреляет из лука, маг кастует фаербол.

````python
class MeleeEnemy:
    def attack(self): return "Меч!"

class RangedEnemy:
    def attack(self): return "Стрела!"

class MagicEnemy:
    def attack(self): return "Фаербол!"

for e in [MeleeEnemy(), RangedEnemy(), MagicEnemy()]:
    print(e.attack())
````

# 10. Магические методы

## 10.1. Что такое магические методы

**Магические методы** (dunder methods, special methods) — методы с именами, начинающимися и заканчивающимися двойным подчёркиванием (`__method__`). Они позволяют объектам **интегрироваться с синтаксисом Python**: операторами, встроенными функциями, циклами.

Python автоматически вызывает магические методы в определённых ситуациях:

| Синтаксис | Метод |
|---|---|
| `print(obj)` | `__str__` |
| `repr(obj)` | `__repr__` |
| `obj1 + obj2` | `__add__` |
| `obj1 == obj2` | `__eq__` |
| `len(obj)` | `__len__` |
| `obj[key]` | `__getitem__` |
| `for x in obj` | `__iter__` |
| `hash(obj)` | `__hash__` |

## 10.2. `__str__` и `__repr__`

Эти два метода отвечают за строковое представление объекта:

- **`__str__`** — читаемое представление для пользователя. Вызывается `print()` и `str()`.
- **`__repr__`** — однозначное представление для отладки. Вызывается `repr()` и используется в контейнерах (списках, словарях).

**Правило:** `__repr__` должен быть однозначным — в идеале `eval(repr(obj))` должен создавать эквивалентный объект. `__str__` должен быть читаемым для человека. Если `__str__` не определён, Python использует `__repr__`.

## 10.3. `__eq__`, `__hash__`, `__lt__`

Магические методы позволяют **сравнивать** объекты:

- `__eq__` — равенство (`==`).
- `__hash__` — хеш (для использования в set/dict).
- `__lt__`, `__gt__`, `__le__`, `__ge__` — сравнения для сортировки.

**Важно:** если вы определяете `__eq__`, вы **должны** определить и `__hash__`, иначе объекты станут нехешируемыми и не смогут использоваться в множествах и словарях. Это частая ловушка.

## 10.4. Арифметические операторы

`__add__`, `__sub__`, `__mul__` и другие позволяют объектам вести себя как числа. Классический пример — `Vector2`, где сложение векторов и умножение на скаляр выглядят естественно.

**Тонкость:** для `3 * v` (когда слева не объект) нужен `__rmul__` (right multiply). Python сначала пробует `int.__mul__(3, v)`, получает `NotImplemented` и вызывает `v.__rmul__(3)`.

**Для углублённого изучения:**
- [Real Python: Magic Methods](https://realpython.com/python-magic-methods/)
- [Python Data Model](https://docs.python.org/3/reference/datamodel.html)

````python
class Vector2:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __str__(self):
        return f"({self.x}, {self.y})"

    def __repr__(self):
        return f"Vector2({self.x}, {self.y})"

    def __eq__(self, other):
        return isinstance(other, Vector2) and (self.x, self.y) == (other.x, other.y)

    def __hash__(self):
        return hash((self.x, self.y))

    def __add__(self, other):
        return Vector2(self.x + other.x, self.y + other.y)

    def __mul__(self, k):
        return Vector2(self.x * k, self.y * k)

    def __rmul__(self, k):
        return self * k

    def __lt__(self, other):
        return (self.x**2 + self.y**2) < (other.x**2 + other.y**2)


v1, v2 = Vector2(1, 2), Vector2(3, 4)
print(v1, repr(v1))
print(v1 + v2)
print(3 * v1)
print(v1 == Vector2(1, 2))
print(sorted([v2, v1]))
````

# 11. MRO — порядок разрешения методов

## 11.1. Что такое MRO

**MRO** (Method Resolution Order) — порядок, в котором Python ищет атрибуты и методы в иерархии классов. Когда вы вызываете `obj.method()`, Python просматривает классы в порядке MRO, начиная с самого производного.

## 11.2. Алгоритм C3-линеаризации

Python использует алгоритм **C3-линеаризации**, который гарантирует:
1. **Локальный порядок приоритетов** сохраняется (порядок родителей в объявлении класса).
2. **Порядок родителей** сохраняется (монотонность).
3. **Единственность** — каждый класс в MRO встречается один раз.

Если эти условия невыполнимы — Python выбрасывает `TypeError: Cannot create a consistent MRO`. Это редкая ситуация, но она возможна при запутанных иерархиях.

## 11.3. Просмотр MRO

MRO можно посмотреть через атрибут `__mro__` или метод `mro()`:

```python
class D(B, C): ...
print(D.__mro__)
# (D, B, C, A, object)
```

## 11.4. Кооперативное наследование

При множественном наследовании важно, чтобы **все классы вызывали `super()`**. Это называется **кооперативным наследованием** — каждый класс явно передаёт управление следующему в MRO. Если хотя бы один класс не вызовет `super()`, цепочка прервётся.

**Для углублённого изучения:**
- [Wikipedia: C3-линеаризация](https://ru.wikipedia.org/wiki/C3-линеаризация)
- [Guido van Rossum: Method Resolution Order](https://www.python.org/download/releases/2.3/mro/)

````python
class A:
    def who(self): return "A"

class B(A):
    def who(self): return "B"

class C(A):
    def who(self): return "C"

class D(B, C):
    pass

print([c.__name__ for c in D.__mro__])
print(D().who())
````

# 12. Композиция vs наследование

## 12.1. Проблема наследования

Наследование — мощный инструмент, но его легко **злоупотребить**. Классический пример: игровой персонаж может быть воином, магом, лучником. Если использовать наследование, мы получим комбинаторный взрыв:

```
Character
├── Warrior
├── Mage
├── Archer
├── WarriorMage
├── WarriorArcher
├── MageArcher
└── WarriorMageArcher
```

Количество подклассов растёт **экспоненциально** с числом признаков. Кроме того, наследование создаёт **жёсткую связь**: изменение родителя может сломать потомков.

## 12.2. Композиция: альтернативный подход

**Композиция** — подход, при котором объект **содержит** другие объекты, а не наследуется от них. Вместо «воин **является** персонажем» используется «персонаж **имеет** класс воина».

**Отношение «имеет» (has-a):**  
`Character` **имеет** `char_class`. Это более гибко: можно менять класс персонажа во время игры, комбинировать способности, добавлять новые классы без изменения иерархии `Character`.

## 12.3. Правило проектирования

**«Предпочитайте композицию наследованию»** — один из ключевых принципов, сформулированных в книге «Design Patterns» (Gang of Four).

Это не означает, что наследование бесполезно. Это означает, что **наследование следует использовать только тогда**, когда:
- отношение действительно «is-a»;
- полиморфизм необходим;
- иерархия стабильна и не растёт экспоненциально.

## 12.4. Практический пример: игровые компоненты

Класс `GameObject` собирается из компонентов: `Position`, `Health`, `Renderable`. Каждый компонент отвечает за одну аспект (координаты, здоровье, отрисовку). Это упрощённая версия паттерна **Component**, который будет подробно изучен в модуле 4.

**Для углублённого изучения:**
- [CodeGym: Composition vs Inheritance](https://codegym.cc/groups/posts/composition-vs-inheritance-in-python)
- [Design Patterns (GoF): Favor Composition over Inheritance](https://en.wikipedia.org/wiki/Composition_over_inheritance)

````python
class Position:
    def __init__(self, x=0, y=0):
        self.x, self.y = x, y
    def move(self, dx, dy):
        self.x += dx; self.y += dy
    def __repr__(self):
        return f"({self.x},{self.y})"

class Health:
    def __init__(self, max_hp=100):
        self.max = max_hp
        self.current = max_hp
    def damage(self, amount):
        self.current = max(0, self.current - amount)

class GameObject:
    def __init__(self, name):
        self.name = name
        self.position = Position()   # композиция
        self.health = Health()       # композиция
    def move(self, dx, dy):
        self.position.move(dx, dy)
    def __repr__(self):
        return f"{self.name} {self.position} hp={self.health.current}"

obj = GameObject("Player")
obj.move(10, 20)
obj.health.damage(30)
print(obj)
````

# 13. Типичные ошибки

## 13.1. «Божественный объект» (God Object)

Класс, который «знает всё и делает всё». Например, `Game` с методами `load_graphics`, `update_players`, `check_collisions`, `save_game` и ещё пятьюдесятью. Нарушает **принцип единственной ответственности** (SRP), который будет изучен в модуле 3. Сложен для понимания, тестирования и изменения.

**Решение:** разделить ответственности между несколькими классами: `Renderer`, `InputHandler`, `PhysicsEngine`, `AudioManager`, `GameState`.

## 13.2. Анемичная модель предметной области (Anemic Domain Model)

Противоположная крайность: классы содержат только данные (геттеры/сеттеры), а вся логика вынесена в отдельные функции. Это **антипаттерн**, потому что логика, естественно принадлежащая объекту, оказывается «снаружи». ООП предполагает, что объект инкапсулирует и данные, и поведение.

**Решение:** перенести логику внутрь объектов. Вместо `damage_player(player, amount)` — `player.take_damage(amount)`.

## 13.3. Наследование ради переиспользования кода

Использование наследования только для того, чтобы не дублировать код, даже если отношение «is-a» не выполняется. Например, `Player(Logger)` — игрок **не является** логгером. Правильное решение — композиция: `self.logger = logger`.

## 13.4. Изменяемые значения по умолчанию

Использование `[]` или `{}` как default-аргументов в `__init__`. Приводит к неожиданному «шарингу» состояния между объектами.

**Решение:** `items=None` + `if items is None: items = []`.

## 13.5. Затенение атрибутов класса

Присвоение `obj.attr = value` создаёт атрибут экземпляра, а не изменяет атрибут класса. Это частая ошибка при работе со счётчиками и константами.

**Решение:** изменяйте через `ClassName.attr`.

## 13.6. Игнорирование `super()`

Прямой вызов `ParentClass.__init__(self, ...)` вместо `super().__init__(...)`. Ломает кооперативное наследование и MRO.

**Решение:** всегда используйте `super()`.
